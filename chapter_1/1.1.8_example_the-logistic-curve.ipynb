# 1.1.8 Example: The Logistic Curve

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Grid configuration for vector field
t = np.linspace(-10, 5, 18)
x = np.linspace(-0.5, 1.5, 18)
T, X = np.meshgrid(t, x)

# 2. ODE Vectors Components: x' = x(1 - x)
dT = np.ones_like(T)  # Time's variation is uniform
dX = X * (1 - X)

# Normalization for all vectors that has equal sizes
magnitude = np.sqrt(dT**2 + dX**2)
dT_norm = dT / magnitude
dX_norm = dX / magnitude

# 3. Preparation of integral curves
t_sol = np.linspace(-10, 5, 200)
K_values = np.linspace(-5.0, 5.0, 20)

# --- Criation of Graph ---
plt.figure(figsize=(9, 6))

# Direction field
plt.quiver(T, X, dT_norm, dX_norm, color='gray', alpha=0.6,
           angles='xy', scale=22, label='Vectors (Direction Field)')

# Integral curves for such constant K
for K in K_values:
    x_sol = (K * np.exp(t_sol)) / (1 + K * np.exp(t_sol))

    x_sol[np.abs(1 + K * np.exp(t_sol)) < 0.05] = np.nan

    plt.plot(t_sol, x_sol, label=f'(K={K:.2f})', linewidth=2)

# Balance Lines (Where x' = 0)
plt.axhline(1, color='green', linestyle='--', linewidth=1.5, label='Stable Balance (x = 1)')
plt.axhline(0, color='red', linestyle=':', linewidth=1.5, label='Unstable Balance (x = 0)')

# Additional visual configurations
plt.title("Direction Field and Integral Curve: $x' = x(1-x)$", fontsize=14)
plt.xlabel("Time ($t$)", fontsize=12)
plt.ylabel("State ($x$)", fontsize=12)
plt.xlim(-10, 5)
plt.ylim(-0.5, 1.5)
plt.grid(True, alpha=0.3)
plt.legend(loc='upper left', bbox_to_anchor=(1, 1))
plt.tight_layout()

# Show the graph
plt.show()
```
