# 1.1.14 Example: Free Fall (V. I. Arnold — *Ordinary Differential Equations*)

This notebook implements the scientific and geometric **dual-panel 3D visualization** for the free fall problem, strictly adhering to the geometric framework established by **Vladimir I. Arnold** in his landmark textbook *Ordinary Differential Equations* (Section 1.1: *Phase Spaces*).

---

### 1. Dynamical System Formulation
Newton's equation of motion for a point mass $m$ under uniform gravitational acceleration $g$ is:
$$ m \ddot{x} = -mg \implies \ddot{x} = -g $$

Defining state variables in the phase space $\mathbb{R}^2$:
- $x_1 = x$ (position / height);
- $x_2 = \dot{x}_1$ (velocity).

We obtain the first-order autonomous system of ordinary differential equations:
$$\begin{cases} \dot{x}_1 = x_2 \\ \dot{x}_2 = -g \end{cases}$$

The associated phase velocity vector field on $\mathbb{R}^2$ is $V(x_1, x_2) = (x_2, -g)$.

---

### 2. First Integral and Mechanical Energy Conservation
Multiplying the second equation by $m x_2 = m \dot{x}_1$, we obtain:
$$ m x_2 \dot{x}_2 + mg \dot{x}_1 = 0 \implies \frac{d}{dt} \left( \frac{1}{2}m x_2^2 + mg x_1 \right) = 0 $$

The total mechanical energy function:
$$ E(x_1, x_2) = \frac{1}{2}m x_2^2 + mg x_1 $$
is a **first integral** (conserved quantity / motion invariant) along any phase trajectory:
$$ L_V E = \nabla E \cdot V = (mg)(x_2) + (m x_2)(-g) \equiv 0 $$

For the physical parameters prescribed:
- $g = 9.8\ \text{m/s}^2$
- $m = 1.0\ \text{kg}$
- Initial conditions at $t = 0$: $x_1(0) = 0.0\ \text{m}$, $x_2(0) = 14.0\ \text{m/s}$
- Constant Mechanical Energy: $E_0 = \frac{1}{2}(1.0)(14.0)^2 + (1.0)(9.8)(0.0) = 98.0\ \text{J}$
- Time interval: $t \in [0, 3]\ \text{s}$

The phase trajectories are the level curves given by $E(x_1, x_2) = E_0$, which defines the parabola:
$$ x_1 = \frac{E_0}{mg} - \frac{1}{2g} x_2^2 = 10.0 - \frac{x_2^2}{19.6} $$

---

### 3. Geometry in the Dual 3D Panels

1. **Panel 1: Extended Phase Space $\mathbb{R} \times \mathbb{R}^2$ with coordinates $(t, x_1, x_2)$**
   - **Invariant Manifold (Parabolic Cylinder)**: The energy constraint $E(x_1, x_2) = E_0$ extended across all $t \in [0, 3]$ generates a semitransparent parabolic cylinder.
   - **3D Integral Curve**: $\gamma(t) = \left(t, -\frac{1}{2}gt^2 + v_0 t + x_0, -gt + v_0\right)$ highlighted in bold red, lying directly on the parabolic cylinder surface.
   - **Extended 3D Vector Field**: $\tilde{V}(t, x_1, x_2) = (1, x_2, -g)$ represented by 3D arrows/cones everywhere tangent to the cylindrical manifold.
   - **Phase Plane Projection at $t = 0$**: Orthogonal shadow of the trajectory on the lateral wall, accompanied by the 2D autonomous vector field $V(x_1, x_2) = (x_2, -g)$ (pointing right and down for $x_2 > 0$, purely downward at $x_2 = 0$, and left and down for $x_2 < 0$) and orthogonal dashed drop lines.

2. **Panel 2: First Integral Relief with coordinates $(x_1, x_2, E)$**
   - **Energy Surface**: The 3D graph $E(x_1, x_2) = \frac{1}{2}m x_2^2 + mg x_1$ forming an inclined parabolic trough.
   - **Constant Energy Plane**: Horizontal semitransparent cutting plane at height $E = E_0 = 98.0\ \text{J}$.
   - **Level Curve and Base Projection**: The 3D intersection curve $E(x_1, x_2) = E_0$ and its vertical orthogonal projection onto the base plane $E = 0$ alongside the vector field $V(x_1, x_2) = (x_2, -g)$, geometrically demonstrating that phase trajectories are the level curves of mechanical energy.
